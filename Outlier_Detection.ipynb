Task name: Outlier Detection Using Box Plot
Objective: Identify potential outliers in a numerical dataset using a box plot.
Tools: Python, Pandas, Matplotlib, Jupyter Notebook
Deliverables:
- Box plot visualization
- Identification of potential outliers
- Short explanation of findings

```python

import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python

# Employee Salary Dataset

df = pd.DataFrame({
    "Employee": [
        "A", "B", "C", "D", "E",
        "F", "G", "H", "I", "J"
    ],
    "Salary": [
        25000, 28000, 30000, 32000, 35000,
        38000, 40000, 42000, 45000, 150000
    ]
})

# Display the dataset
print(df)
```

```python

print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
print(df["Salary"].describe())
```

```python

# Calculate the first and third quartiles
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

# Calculate the Interquartile Range
IQR = Q3 - Q1

# Calculate lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python

# Identify values outside the calculated bounds
outliers = df[
    (df["Salary"] < lower_bound) |
    (df["Salary"] > upper_bound)
]

print("Potential Outliers:")
print(outliers)

if outliers.empty:
    print("No potential outliers detected.")
else:
    print("Number of potential outliers:", len(outliers))
```
